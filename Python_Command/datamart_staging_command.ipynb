```python

import sqlite3
from pathlib import Path
import pandas as pd

# Detect project root
current_folder = Path.cwd()

if current_folder.name.lower() == "python_command":
    project_root = current_folder.parent
else:
    project_root = current_folder

dim_database_path = (
    project_root
    / "Datamart"
    / "Dim"
    / "dim_database.db"
)

staging_database_path = (
    project_root
    / "Datamart"
    / "Staging"
    / "staging_bank_statement_2026.db"
)

print("DIM database:", dim_database_path.resolve())
print("Staging database:", staging_database_path.resolve())
```

```python
with sqlite3.connect(staging_database_path) as connection:
    staging_bank_df = pd.read_sql_query(
        """
        SELECT *
        FROM staging_bank_statement_2026
        """,
        connection
    )

print("Staging shape:", staging_bank_df.shape)
display(staging_bank_df.head())
```

```python
with sqlite3.connect(dim_database_path) as connection:
    dim_account = pd.read_sql_query(
        "SELECT * FROM dim_account",
        connection
    )

    dim_transaction_type = pd.read_sql_query(
        "SELECT * FROM dim_transaction_type",
        connection
    )

    dim_counterparty_role = pd.read_sql_query(
        "SELECT * FROM dim_counterparty_role",
        connection
    )

    dim_customer_name_confidence = pd.read_sql_query(
        "SELECT * FROM dim_customer_name_confidence",
        connection
    )
```

```python
print("dim_account")
display(dim_account.head())

print("dim_transaction_type")
display(dim_transaction_type.head())

print("dim_counterparty_role")
display(dim_counterparty_role.head())

print("dim_customer_name_confidence")
display(dim_customer_name_confidence.head())
```

```python
staging_bank_df["account_number"] = (
    staging_bank_df["account_number"]
    .astype("string")
    .str.strip()
)

dim_account["account_number"] = (
    dim_account["account_number"]
    .astype("string")
    .str.strip()
)

staging_bank_df["transaction_type_standard"] = (
    staging_bank_df["transaction_type_standard"]
    .astype("string")
    .str.strip()
)

dim_transaction_type["transaction_type_name"] = (
    dim_transaction_type["transaction_type_name"]
    .astype("string")
    .str.strip()
)

staging_bank_df["counterparty_role"] = (
    staging_bank_df["counterparty_role"]
    .astype("string")
    .str.strip()
)

dim_counterparty_role["counterparty_role_name"] = (
    dim_counterparty_role["counterparty_role_name"]
    .astype("string")
    .str.strip()
)

staging_bank_df["customer_name_confidence"] = (
    staging_bank_df["customer_name_confidence"]
    .astype("string")
    .str.strip()
)

dim_customer_name_confidence["customer_name_confidence_name"] = (
    dim_customer_name_confidence["customer_name_confidence_name"]
    .astype("string")
    .str.strip()
)
```

```python
mapped_bank_df = staging_bank_df.copy()

# Map account dimension
mapped_bank_df = mapped_bank_df.merge(
    dim_account[
        [
            "account_key",
            "account_number",
            "branch",
            "account_type"
        ]
    ],
    on="account_number",
    how="left",
    validate="many_to_one"
)

# Map transaction type dimension
mapped_bank_df = mapped_bank_df.merge(
    dim_transaction_type[
        [
            "transaction_type_key",
            "transaction_type_name"
        ]
    ],
    left_on="transaction_type_standard",
    right_on="transaction_type_name",
    how="left",
    validate="many_to_one"
)

# Map counterparty role dimension
mapped_bank_df = mapped_bank_df.merge(
    dim_counterparty_role[
        [
            "counterparty_role_key",
            "counterparty_role_name"
        ]
    ],
    left_on="counterparty_role",
    right_on="counterparty_role_name",
    how="left",
    validate="many_to_one"
)

# Map confidence dimension
mapped_bank_df = mapped_bank_df.merge(
    dim_customer_name_confidence[
        [
            "customer_name_confidence_key",
            "customer_name_confidence_name"
        ]
    ],
    left_on="customer_name_confidence",
    right_on="customer_name_confidence_name",
    how="left",
    validate="many_to_one"
)

print("Mapped shape:", mapped_bank_df.shape)
display(mapped_bank_df.head())
```

```python
print("Unmatched account keys:")
display(
    mapped_bank_df.loc[
        mapped_bank_df["account_key"].isna(),
        ["bank", "account_number"]
    ]
    .drop_duplicates()
    .sort_values(["bank", "account_number"])
)
```

```python
final_columns = [
    "bank",
    "account_key",
    "account_number",
    "branch",
    "account_type",
    "transaction_date",
    "transaction_time",
    "withdrawal",
    "deposit",
    "outstanding_balance",
    "transaction_type_key",
    "transaction_type",
    "transaction_type_standard",
    "description_clean",
    "customer_name",
    "counterparty_role_key",
    "counterparty_role",
    "customer_name_confidence_key",
    "customer_name_confidence"
]

mapped_bank_df = mapped_bank_df[
    [
        column
        for column in final_columns
        if column in mapped_bank_df.columns
    ]
].copy()

display(mapped_bank_df.head())
print(mapped_bank_df.dtypes)
print("Final shape:", mapped_bank_df.shape)
```

```python
from pathlib import Path
import pandas as pd

# Detect project root
current_folder = Path.cwd()

if current_folder.name.lower() == "python_command":
    project_root = current_folder.parent
else:
    project_root = current_folder

# Export folder
export_staging_folder = (
    project_root
    / "Export_CSV"
    / "Staging"
)

export_staging_folder.mkdir(
    parents=True,
    exist_ok=True
)

# Final staging columns
final_columns = [
    "bank",
    "account_key",
    "branch",
    "account_type",
    "transaction_date",
    "transaction_time",
    "withdrawal",
    "deposit",
    "outstanding_balance",
    "transaction_type_key",
    "description_clean",
    "customer_name",
    "counterparty_role_key",
    "customer_name_confidence_key",
]

staging_export_df = mapped_bank_df[
    [
        column
        for column in final_columns
        if column in mapped_bank_df.columns
    ]
].copy()

# Ensure transaction_date is datetime
staging_export_df["transaction_date"] = pd.to_datetime(
    staging_export_df["transaction_date"],
    errors="coerce"
)

# Export one CSV per year
for year, year_df in staging_export_df.groupby(
    staging_export_df["transaction_date"].dt.year
):
    year = int(year)

    output_path = (
        export_staging_folder
        / f"staging_bank_statement_{year}.csv"
    )

    year_df = year_df.copy()

    year_df["transaction_date"] = (
        year_df["transaction_date"]
        .dt.strftime("%Y-%m-%d")
    )

    year_df.to_csv(
        output_path,
        index=False,
        encoding="utf-8-sig"
    )

    print(f"Year: {year}")
    print(f"Rows exported: {len(year_df):,}")
    print(f"CSV: {output_path.resolve()}")
```
