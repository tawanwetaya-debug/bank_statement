```python
import pandas as pd
import numpy as np
import re

from pathlib import Path
from IPython.display import display


# Notebook อยู่ใน:
# Bank_Account_Project/Python_Command/KBANK.ipynb

notebook_folder = Path.cwd()
project_folder = notebook_folder.parent

kbank_folder = (
    project_folder
    / "Year_End_2026"
    / "KBANK_Account"
)

output_folder = project_folder / "Datamart"
output_folder.mkdir(parents=True, exist_ok=True)

supported_extensions = {".csv", ".xls", ".xlsx"}

kbank_files = sorted(
    file
    for file in kbank_folder.rglob("*")
    if file.is_file()
    and file.suffix.lower() in supported_extensions
    and not file.name.startswith("~$")
)

print("KBANK folder:", kbank_folder)
print("Folder exists:", kbank_folder.exists())
print("Total files found:", len(kbank_files))

for file in kbank_files:
    print(file.relative_to(kbank_folder))
```

```python
def read_kbank_raw(file_path):
    suffix = file_path.suffix.lower()

    if suffix == ".csv":
        encodings = [
            "utf-8-sig",
            "utf-8",
            "cp874",
            "tis-620"
        ]

        last_error = None

        for encoding in encodings:
            try:
                return pd.read_csv(
                    file_path,
                    header=None,
                    dtype=str,
                    encoding=encoding,
                    sep=",",
                    engine="python",
                    on_bad_lines="skip"
                )

            except UnicodeDecodeError as error:
                last_error = error

        raise ValueError(
            f"Cannot decode file: {file_path.name}"
        ) from last_error

    if suffix in {".xls", ".xlsx"}:
        return pd.read_excel(
            file_path,
            header=None,
            dtype=str
        )

    raise ValueError(
        f"Unsupported file type: {file_path.suffix}"
    )
```

```python
def clean_cell(value):
    if pd.isna(value):
        return ""

    return (
        str(value)
        .replace("\n", " ")
        .replace("\r", " ")
        .strip()
    )
```

```python
def extract_account_number(raw):
    account_labels = [
        "เลขที่บัญชีเงินฝาก",
        "เลขที่บัญชี",
        "บัญชีเงินฝาก"
    ]

    account_pattern = re.compile(
        r"\b\d{3}-\d-\d{5}-\d\b"
    )

    # ตรวจเฉพาะส่วนบนของ Statement
    max_rows = min(len(raw), 30)

    for row_index in range(max_rows):
        for col_index in range(len(raw.columns)):
            cell = clean_cell(
                raw.iat[row_index, col_index]
            )

            if not any(
                label in cell
                for label in account_labels
            ):
                continue

            # กรณีข้อความและเลขบัญชีอยู่ในเซลล์เดียวกัน
            match = account_pattern.search(cell)

            if match:
                return match.group(0)

            # ตรวจเซลล์ทางขวา
            for next_col in range(
                col_index + 1,
                min(col_index + 10, len(raw.columns))
            ):
                candidate = clean_cell(
                    raw.iat[row_index, next_col]
                )

                match = account_pattern.search(candidate)

                if match:
                    return match.group(0)

    # สำรอง: ค้นหารูปแบบเลขบัญชีในส่วนบนทั้งหมด
    for row_index in range(max_rows):
        for col_index in range(len(raw.columns)):
            candidate = clean_cell(
                raw.iat[row_index, col_index]
            )

            match = account_pattern.search(candidate)

            if match:
                return match.group(0)

    return None
```

```python
def find_transaction_header(raw):
    for row_index in range(len(raw)):
        row_values = (
            raw.iloc[row_index]
            .fillna("")
            .astype(str)
            .map(clean_cell)
        )

        has_date = row_values.eq("วันที่").any()

        has_time = row_values.str.contains(
            r"เวลา|วันที่รายการ",
            regex=True,
            na=False
        ).any()

        has_balance = row_values.str.contains(
            r"ยอดคงเหลือ",
            regex=True,
            na=False
        ).any()

        has_amount = row_values.str.contains(
            r"ถอนเงิน|ฝากเงิน",
            regex=True,
            na=False
        ).any()

        if has_date and (
            has_time
            or has_balance
            or has_amount
        ):
            return row_index

    return None
```

```python
def normalize_kbank_column(column_name):
    column_name = clean_cell(column_name)

    replacements = {
        "วันที่": "transaction_date",
        "เวลา/ วันที่รายการ": "transaction_time_description",
        "เวลา/วันที่รายการ": "transaction_time_description",
        "เวลา / วันที่รายการ": "transaction_time_description",
        "ถอนเงิน": "withdrawal",
        "ฝากเงิน": "deposit",
        "ยอดคงเหลือ": "outstanding_balance",
        "ช่องทาง": "channel",
        "รายละเอียด": "description"
    }

    return replacements.get(
        column_name,
        column_name
    )
```

```python
def make_unique_columns(columns):
    unique_columns = []
    column_count = {}

    for index, column in enumerate(columns):
        column = normalize_kbank_column(column)

        if not column:
            column = f"extra_column_{index + 1}"

        if column in column_count:
            column_count[column] += 1
            column = (
                f"{column}_{column_count[column]}"
            )
        else:
            column_count[column] = 0

        unique_columns.append(column)

    return unique_columns
```

```python
def read_kbank_statement(file_path, root_folder):
    raw = read_kbank_raw(file_path)

    raw = raw.map(clean_cell)

    account_number = extract_account_number(raw)
    header_row = find_transaction_header(raw)

    if header_row is None:
        raise ValueError(
            f"ไม่พบแถวหัวตารางที่มีคำว่า วันที่: "
            f"{file_path.name}"
        )

    header_values = raw.iloc[header_row].tolist()

    # หาคอลัมน์เริ่มต้นที่มีคำว่า วันที่
    start_column = None

    for column_index, value in enumerate(
        header_values
    ):
        if clean_cell(value) == "วันที่":
            start_column = column_index
            break

    if start_column is None:
        raise ValueError(
            f"ไม่พบคอลัมน์ วันที่: {file_path.name}"
        )

    # เก็บตั้งแต่คอลัมน์ วันที่ ไปจนถึงขวาสุด
    transaction_area = raw.iloc[
        header_row:,
        start_column:
    ].copy()

    # ลบเฉพาะคอลัมน์ที่ว่างทั้งหมด
    valid_columns = transaction_area.apply(
        lambda column: (
            column
            .fillna("")
            .astype(str)
            .str.strip()
            .ne("")
            .any()
        )
    )

    transaction_area = transaction_area.loc[
        :,
        valid_columns
    ]

    columns = make_unique_columns(
        transaction_area.iloc[0].tolist()
    )

    df = transaction_area.iloc[1:].copy()
    df.columns = columns
    df = df.reset_index(drop=True)

    df = df.replace(
        {
            "": np.nan,
            "nan": np.nan,
            "NaN": np.nan,
            "None": np.nan
        }
    )

    # ลบแถวที่ว่างทั้งหมด
    df = df.dropna(how="all").copy()

    # เก็บเฉพาะแถว transaction ที่มีวันที่
    if "transaction_date" in df.columns:
        df = df[
            df["transaction_date"].notna()
        ].copy()

    relative_path = file_path.relative_to(
        root_folder
    )

    month_folder = (
        relative_path.parts[0]
        if len(relative_path.parts) > 1
        else None
    )

    # เพิ่มข้อมูลจากส่วนบนของ Statement
    df.insert(0, "bank", "KBANK")
    df.insert(1, "account_number", account_number)
    df.insert(2, "month_folder", month_folder)
    df.insert(3, "source_file", file_path.name)
    df.insert(
        4,
        "source_path",
        str(relative_path)
    )

    return df
```

```python
all_kbank_data = []
failed_files = []

for file in kbank_files:
    try:
        file_df = read_kbank_statement(
            file_path=file,
            root_folder=kbank_folder
        )

        all_kbank_data.append(file_df)

        account = (
            file_df["account_number"].iloc[0]
            if not file_df.empty
            else None
        )

        print(
            f"OK | "
            f"{file.relative_to(kbank_folder)} | "
            f"Account: {account} | "
            f"Rows: {len(file_df):,}"
        )

    except Exception as error:
        failed_files.append({
            "source_file": str(
                file.relative_to(kbank_folder)
            ),
            "error": str(error)
        })

        print(
            f"FAILED | "
            f"{file.relative_to(kbank_folder)} | "
            f"{error}"
        )
```
